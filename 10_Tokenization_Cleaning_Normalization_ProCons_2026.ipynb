- **Stemming** usually refers to a crude heuristic process that chops off the ends of words in the hope of achieving this goal correctly most of the time, and often includes the **<em>removal of derivational affixes</em>**.

- **Lemmatization** usually refers to doing things properly with the use of a vocabulary and morphological analysis of words, normally aiming to **remove inflectional endings** only and to return the **base or dictionary form** of a word, which is known as the lemma.

    - If confronted with the token '**saw**', stemming might return just '**s**', whereas lemmatization would attempt to return either '**see**'(verb) or '**saw**'(noun) depending on whether the use of the token was as a _verb_ or a _noun_.

    - Linguistic processing for stemming or lemmatization is often done by an additional plug-in component to the indexing process, and a number of such components exist, both commercial and open-source. The most common algorithm for **stemming English**, and one that has repeatedly been shown to be empirically very effective, is <font color = 'red'> **Porter's algorithm (Porter, 1980)**.

    - You can use a **lemmatizer**, a tool from Natural Language Processing which does full morphological analysis to <em>accurately identify the lemma for each word</em>. <font color = 'blue'> Doing full morphological analysis produces at most very modest benefits for retrieval</font>. It is hard to say more, because either form of normalization tends not to improve English information retrieval performance in aggregate - at least not by very much. While it helps a lot for some queries, it equally hurts performance a lot for others. <font color = 'red'> Stemming increases recall while harming precision.</font>

>The **Porter stemmer** stems all of the following words:
>
>>'operate', 'operating, ' operates', ' operation', ' operative', ' operatives', ' operational'
$\Rightarrow$ '**oper**'.
>
> Defect $\Rightarrow$ We lose considerable precision on queries such as the following with Porter stemming:
>> "oper<font color = 'blue'>**ational**</font> and research", "oper<font color = 'blue'>**ating**</font> and system", "oper<font color = 'blue'>**ative**</font> and dentistry"

For a case like this, moving to using a lemmatizer would not completely fix the problem because **particular inflectional forms are used in particular <font color = 'blue'>collocations</font>**: a sentence with the words operate and system is not a good match for the query operating and system. Getting better value from term normalization depends more on pragmatic issues of word use than on formal issues of linguistic morphology. 특정 굴절형이 특정한 연어(collocations)에서 사용되기 때문에 'operate'와 'system'이라는 단어가 포함된 문장은 'operating'과 'system'으로 이루어진 질의에는 좋은 매칭이 되지 않는다. 용어 정규화(term normalization)를 통해 더 나은 가치를 얻는 것은 언어 형태론적인 형식적인 문제보다는 단어 사용의 화용론적인(실용적인) 문제에 더 많이 달려있다.

- Example:

- Uses vocabulary + morphological analysis

- Produces real dictionary words

- Example:

Stopwords are the English words which does not add much meaning to a sentence. They can safely be ignored without sacrificing the meaning of the sentence. For example, the words like the, he, have etc. Such words are already captured this in corpus named corpus. We first download it to our python environment.

Stop words are common words like ‘the’, ‘and’, ‘I’, etc. that are very frequent in text, and so don’t convey insights into the specific topic of a document. We can remove these stop words from the text in a given corpus to clean up the data, and identify words that are more rare and potentially more relevant to what we’re interested in.

Text may contain stop words like ‘the’, ‘is’, ‘are’. Stop words can be filtered from the text to be processed. There is no universal list of stop words in nlp research, however the nltk module contains a list of stop words.

# **Conclusion:**

Tokenization cannot be simply regarded as splitting a corpus by whitespace after removing punctuation.

Punctuation marks or special characters should not be removed indiscriminately. When filtering words from a given corpus, it is not appropriate to simply discard punctuation or special symbols. During corpus preprocessing, even punctuation can sometimes be classified as a token. For a basic example, a period (.) helps identify sentence boundaries, so it may not be excluded when extracting words.

There are also cases where punctuation is part of a word itself, such as m.p.h, Ph.D, or AT&T. As for special characters like the dollar sign or slash (/), $45.55 represents a price, and 01/02/06 represents a date. In such cases, we may want to treat 45.55 as a single unit rather than separating it into 45 and 55.

Commas (,) may also appear within numbers. When expressing numerical values, commas are often inserted every three digits, as in 123,456,789.

There are also abbreviations and words that contain spaces. In tokenization, the apostrophe (') in English sometimes functions to expand contracted forms. For example, what're is a contraction of what are, and we're is a contraction of we are. In these examples, 're is called a clitic. That is, it refers to the form that appears when a word is used in a contracted form. For instance, in I'm, which is a shortened form of I am, 'm is a clitic.

Consider the words New York and rock 'n' roll. These are single expressions but contain spaces within them. Depending on the purpose, even when a word contains spaces, it may need to be treated as a single token. Therefore, a tokenization process must be capable of recognizing such expressions as one token.
